**Question:**

Create a class CountDown that takes a number n and counts down to 1.
Implement the iterator protocol (__iter__() and __next__() methods).
Use it in a for loop to print numbers in reverse order.

**Answer is Shown below:**

```python
class CountDown:
    def __init__(self, n):
        self.n = n
    def __iter__(self):
        return self

    def __next__(self):
        if self.n <= 0:
            raise StopIteration
        number = self.n
        self.n -= 1
        return number

count = CountDown(5)
for number in count:
    print(number)
```

Create a custom iterator FileLineReader that takes a filename and reads one line at a time.
Implement __iter__() and __next__() so it can be used in a for loop.
Test it by reading from a sample .txt file and printing each line

**Answer is shown below:**

```python
text = """My name is Bayo Nyangskebrifun.
I am learning Python.
I am learning about iterators.
Python is interesting.
"""

with open("sample.txt", "w") as file:
    file.write(text)
```

```python
class FileLineReader:
    def __init__(self, filename):
        self.file = open(filename, "r")

    def __iter__(self):
        return self

    def __next__(self):
        line = self.file.readline()

        if line == "":
            self.file.close()
            raise StopIteration

        return line.strip()


reader = FileLineReader("sample.txt")

for line in reader:
    print(line)
```

**Question:**

Create an abstract class Shape with an abstract method area().

Create subclasses Rectangle and Circle that implement the area() method.

Create a list of different shapes and loop through them, printing their area without checking their type.


**Answer is shown below:**


```python
from abc import ABC, abstractmethod


class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2

shapes = [
    Rectangle(10, 5),
    Circle(7),
    Rectangle(6, 4)
]

for shape in shapes:
    print("Area:", shape.area())
```

```python
class EvenNumbers:
    def __init__(self, limit):
        self.limit = limit
        self.number = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.number > self.limit:
            raise StopIteration

        even_number = self.number
        self.number += 2

        return even_number


even_numbers = EvenNumbers(10)

for number in even_numbers:
    print(number)
```

**Question:**

Create a class Fibonacci that generates Fibonacci numbers up to a given n using an iterator protocol.

```python
class Fibonacci:
    def __init__(self, n):
        self.n = n
        self.a = 0
        self.b = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.a > self.n:
            raise StopIteration

        current = self.a
        self.a = self.b
        self.b = current + self.b

        return current


fib = Fibonacci(20)

for number in fib:
    print(number)
```

5. Track which books are available and which are borrowed.

**Answer:**

```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
        self.is_available = True

    def get_description(self):
        return f"{self.title} by {self.author} ({self.year})"

class EBook(Book):
    def __init__(self, title, author, year, file_size, format):
        super().__init__(title, author, year)
        self.file_size = file_size
        self.format = format

    def get_description(self):
        return f"EBook: {self.title} by {self.author}, {self.file_size}MB, {self.format}"

class PrintedBook(Book):
    def __init__(self, title, author, year, weight, dimensions):
        super().__init__(title, author, year)
        self.weight = weight
        self.dimensions = dimensions

    def get_description(self):
        return f"Printed Book: {self.title} by {self.author}, {self.weight}kg, {self.dimensions}"


class Member:
    total_borrowed = 0

    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self._borrowed_books = []

    def borrow_book(self, book):
        if not book.is_available:
            print(f"'{book.title}' is already borrowed.")
            return

        self._borrowed_books.append(book)
        book.is_available = False
        Member.total_borrowed += 1
        print(f"{self.name} borrowed '{book.title}'.")

    def __iter__(self):
        return iter(self._borrowed_books)

    def get_borrowed_books(self):
        return self._borrowed_books

class StudentMember(Member):
    borrowing_limit = 2

    def borrow_book(self, book):
        if len(self._borrowed_books) >= self.borrowing_limit:
            print(f"{self.name} has reached the student borrowing limit.")
            return

        super().borrow_book(book)

class TeacherMember(Member):
    borrowing_limit = 5

    def borrow_book(self, book):
        if len(self._borrowed_books) >= self.borrowing_limit:
            print(f"{self.name} has reached the teacher borrowing limit.")
            return

        super().borrow_book(book)


class Library:
    def __init__(self, library_name):
        self.__library_name = library_name
        self.books = []
        self.members = []

    def add_book(self, book):
        self.books.append(book)

    def add_member(self, member):
        self.members.append(member)

    def __iter__(self):
        return iter(self.books)

    def search_books(self, search_term):
        results = []

        for book in self.books:
            if (search_term.lower() in book.title.lower() or
                    search_term.lower() in book.author.lower()):
                results.append(book)

        return results

    def get_library_name(self):
        return self.__library_name

    def set_library_name(self, new_name):
        self.__library_name = new_name


# Create books
book1 = EBook(
    "Python Basics",
    "John Smith",
    2024,
    5,
    "PDF"
)

book2 = PrintedBook(
    "Introduction to Physics",
    "James Brown",
    2023,
    1.2,
    "25 x 18 cm"
)

book3 = PrintedBook(
    "Data Science",
    "Mary Jones",
    2025,
    1.5,
    "26 x 19 cm"
)

# Create members
student = StudentMember("Bayo", "ST001")
teacher = TeacherMember("Mr. John", "TC001")


# Create library
library = Library("Community Library")

library.add_book(book1)
library.add_book(book2)
library.add_book(book3)

library.add_member(student)
library.add_member(teacher)


# Display book descriptions
print("BOOKS")
for book in library:
    print(book.get_description())


# Borrow books
print("\nBORROWING BOOKS")

student.borrow_book(book1)
student.borrow_book(book2)

teacher.borrow_book(book3)


# Display borrowed books
print("\nSTUDENT BORROWED BOOKS")

for book in student:
    print(book.title)


print("\nTEACHER BORROWED BOOKS")

for book in teacher:
    print(book.title)


# Search for books
print("\nSEARCH RESULTS")

results = library.search_books("Python")

for book in results:
    print(book.get_description())


# Display availability
print("\nBOOK AVAILABILITY")

for book in library:
    status = "Available" if book.is_available else "Borrowed"
    print(book.title, "-", status)


# Total books borrowed
print("\nTOTAL BOOKS BORROWED:", Member.total_borrowed)

# Library name
print("LIBRARY:", library.get_library_name())
```
