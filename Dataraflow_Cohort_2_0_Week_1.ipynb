**Guessing Game**

Below is a program where the computer randomly selects a number between 1 - 20, and the user must guess it. Give hints if the guess is too high or too low.

```python
import random

number = random.randint(20, 50)

guess = int(input("Guess a number between 20 and 50: "))

while guess != number:

    if guess < number:
        print("Too low")
    else:
        print("Too high")

    guess = int(input("Try again: "))

print("Correct! You guessed the number right.")
```

**Assignment 5**

**File Handling**

Below is a program that accepts a list of student names from the user, saves them in a text file, and then reads and displays them.

```python
names = []

number = int(input("How many students do you want to enter? "))

for i in range(number):
    name = input("Enter student name: ")
    names.append(name)

file = open("students.txt", "w")

for name in names:
    file.write(name + "\n")

file.close()

print("Students saved successfully.")

file = open("students.txt", "r")

print("\nStudents in the file:")

for name in file:
    print(name.strip())

file.close()
```

**Data Types in Python**

In Python, data types define the type of data that can be stored in a variable. Python has several built-in data types, each serving different purposes and operations. The following are the primary data types in Python:

**Numeric Data Types**

Numeric data types represent numbers and include:

**int:** Represents integers, e.g., 5, -10.

**float:** Represents floating-point numbers, e.g., 3.14, -0.001.

**complex:** Represents complex numbers, e.g., 1+2j.


```python
# Example of Numeric Data type
a = 5
b = 3.14
c = 1 + 2j
print(type(a)) # class: 'int'>
print(type(b)) # class: 'float'>
print(type(c)) # class: 'complex'>
```

**Sequence Data Types**

Sequence data types represent ordered collections of items and include:

**str:** Represents strings, e.g., "Hello", 'Python'.

**list:** Represents lists, e.g., [1, 2, 3], ["apple", "banana"].

**tuple:** Represents tuples, e.g., (1, 2, 3), ("apple", "banana").

**range:** Represents ranges, e.g., range(5).



**Mapping Data Type**

Mapping data types represent collections of key-value pairs and include:

**dict:** Represents dictionaries. Example. {"name": "John", "age": 30}.



**Set Data Types**

Set data types represent unordered collections of unique items and include:

**set:** Represents sets, e.g. {1, 2, 3}, {"apple", "banana"}.

**frozenset:** Represents immutable sets, e.g., frozenset([1, 2, 3]).

**Boolean Data Type**

Boolean data types represent truth values and include:

**bool:** Represents boolean values, e.g., True, False.

**Difference Between Local and Global Variables in Programming**

**Local Variables**

Local variables are declared within a specific block of code, such as a function or method. They are accessible only within that block and cease to exist once the block is exited. These variables are typically used for temporary storage or calculations within a limited context.

Example, see the code below:

```python
def calculate_sum():
  x = 10
  y = 20
  return (x + y)
print(calculate_sum())
```

**Global Variables**

Global variables are declared outside of any function or block and can be accessed from any part of the program. They persist throughout the program's execution and are often used to share data across multiple functions or modules. For Example, see code below:

```python
global_var = 100
def add_to_global():
  return global_var + 50
print(add_to_global())
```

Difference Between **'for'** and **'while'** loops.

'for' loop is generally used when the number of iterations is known in advance or when iterating over a sequence (list, tuple, string, range, etc.). It automatically handles initialization, condition checking, and iteration updates internally. Example of 'for' loop is shown below:

```python
for num in range(1, 12,):
  print(num)
```

while Loop

While Loop is preferred when the number of iterations is unknown and depends on a condition being met. It requires manual initialization and update of loop variables, and the loop continues until the condition becomes false. Example of 'while" loop is shown below:

```python
password = "PythonRocks"
user_input = ""

while user_input != password:
  user_input = input("Enter the password: ")

print("Access granted!")
```

A program to check if a string is a palindrome (i.e a word that can give  the same word if the letter are reversed) is written below:
